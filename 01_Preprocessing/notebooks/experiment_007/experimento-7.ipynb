# Experimento 7 - Preencher os valores vazios de todas as colunas com as medianas de seus valores e remover o atributo `Insulin`

Acurácia atingida: $0.6020408163265306$

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../../diabetes_dataset.csv")
```

```python
# colunas com valores ausentes: insulin, skinthickness, bloodpressure, bmi e glucose
colunas_preencher = ["SkinThickness", "BloodPressure", "BMI", "Glucose"]

print("médias antes:")
print(df[colunas_preencher].mean())

medianas = df[colunas_preencher].median()
df[colunas_preencher] = df[colunas_preencher].fillna(medianas)

print("medianas utilizadas:", medianas)

print("médias depois:")
print(df[colunas_preencher].mean())

# remover insulin
df = df.drop(columns=["Insulin"])
```

```python
df.to_csv("diabetes-tratado-ex7.csv", index=False)
print("arquivo exportado com", len(df), "registros")
```
