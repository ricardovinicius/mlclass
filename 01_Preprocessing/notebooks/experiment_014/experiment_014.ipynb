# Experimento 14 - Drop Insulin, imputação por mediana, RobustScaler e variável derivada `Glucose_Age`

````python
import pandas as pd
from sklearn.preprocessing import RobustScaler

df = pd.read_csv("../../diabetes_dataset.csv")
df.shape
````

````python
df = df.drop(columns=['Insulin'])
df.columns.tolist()
````

````python
# derivando o atributo `glucose_bmi` 
df["Glucose_Age"] = df["Glucose"] * df["Age"]
````

````python
medianas = df.median(numeric_only=True)
df = df.fillna(medianas)
df.isna().sum()

# medianas
````

````python
NORMALIZE_COLS = ['Glucose', 'BloodPressure', 'SkinThickness', 'BMI', 'DiabetesPedigreeFunction', 'Glucose_Age']
scaler = RobustScaler()
df[NORMALIZE_COLS] = scaler.fit_transform(df[NORMALIZE_COLS])
df.head()
````

````python
df.to_csv("diabetes-tratado-ex14.csv", index=False)
print("treino exportado com", len(df), "registros e colunas:", list(df.columns))
````
