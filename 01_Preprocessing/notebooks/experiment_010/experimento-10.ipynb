# Experimento 10 - Preencher os valores vazios de todas as colunas com as medianas de seus valores e remover os atributos `Insulin` e `SkinThickness`

Acurácia atingida: $0.5816326530612245$

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../../diabetes_dataset.csv")
```

```python
# colunas com valores ausentes: insulin, bloodpressure, bmi e glucose
colunas_preencher = ["BloodPressure", "BMI", "Glucose"]

print("médias antes:")
print(df[colunas_preencher].mean())

medianas = df[colunas_preencher].median()
df[colunas_preencher] = df[colunas_preencher].fillna(medianas)

print("medianas utilizadas:", medianas)

print("médias depois:")
print(df[colunas_preencher].mean())

# remover insulin e skinthickness
df = df.drop(columns=["Insulin", "SkinThickness"])
```

```python
df.to_csv("diabetes-tratado-ex10.csv", index=False)
print("arquivo exportado com", len(df), "registros")
```
