# Experimento 1 - Remoção de registros que possuem campos vazios

Acurácia atingida: $0.5612244897959183$

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../../diabetes_dataset.csv")
```

```python
total = len(df)
print("total de registros no dataset:", total)

qtd = df.isna().any(axis=1).sum()
print("registros com pelo menos um campo vazio:", qtd)

percent = (qtd/total) * 100
print(f"percentual: {percent:.2f}%")

# a perda de registros para treino sera grande
```

```python
total_antes = len(df)

df = df.dropna()

total_depois = len(df)

print("registros removidos:", total_antes-total_depois)
print("registros restantes:", total_depois)
```

```python
df.to_csv("diabetes-tratado-ex1.csv", index=False)
print("arquivo exportado com", total_depois, "registros")
```
